```python
def imi(close,opens,time=14):  
    df=pd.DataFrame()
    df['Gain'] = np.where(close > opens, close - opens, 0)
    df['Loss'] = np.where(close < opens, opens - close, 0)

    # Set the period for IMI (e.g., 14 days)
    period = time

    # Calculate rolling sums of gains and losses
    df['SumGain'] = df['Gain'].rolling(window=period).sum()
    df['SumLoss'] = df['Loss'].rolling(window=period).sum()

    # Calculate IMI
    df['IMI'] = 100 * (df['SumGain'] / (df['SumGain'] + df['SumLoss']))
    return df['IMI']
```

```python

The given imi function is well-suited for calculating the Intraday Momentum Index (IMI) for 1-minute data, as long as you provide a DataFrame with close and open prices sampled at 1-minute intervals. The only change required is a clarification that the time parameter corresponds to the number of intervals (in this case, minutes), rather than days.

Updated Function for 1-Minute Data
python
Copy code
import pandas as pd
import numpy as np

def imi(close, opens, time=14):  
    """
    Calculate the Intraday Momentum Index (IMI) for 1-minute data.

    Parameters:
        close (pd.Series): Series of close prices.
        opens (pd.Series): Series of open prices.
        time (int): The period for the rolling calculation (default is 14 intervals).

    Returns:
        pd.Series: The IMI values.
    """
    df = pd.DataFrame()
    df['Gain'] = np.where(close > opens, close - opens, 0)
    df['Loss'] = np.where(close < opens, opens - close, 0)

    # Calculate rolling sums of gains and losses
    df['SumGain'] = df['Gain'].rolling(window=time).sum()
    df['SumLoss'] = df['Loss'].rolling(window=time).sum()

    # Calculate IMI
    df['IMI'] = 100 * (df['SumGain'] / (df['SumGain'] + df['SumLoss']))
    return df['IMI']  # Returning only the IMI column
```

```python
import time
import concurrent.futures

# Define the worker function
def fetch_data(symbol, error_symbols):
    try:
        new = atha.intraday_minute_data(exchange_segment="NSE_EQ",
                                        instrument_type='EQUITY',
                                      security_id=symbol)
        name=nifty500.loc[nifty500['pSymbol'] == symbol, 'pTrdSymbol'].values[0]
        a = cha(new , symbol ,name)
        print(f"Symbol: {symbol}, Processed Data: {a}")
        #time.sleep(0.1)  # Add the delay as per your requirement
    except Exception as e:
        print(f"Error fetching data for {symbol}: {e}")
        if "DataFrame constructor not properly called!" in str(e):
            error_symbols.append(symbol)  # Store symbol for retry

# Use ThreadPoolExecutor to run the tasks in parallel
def run_in_threads():
    
    global temp_file
    temp_file=pd.DataFrame()
    symbols_to_retry = list(nifty500['pSymbol'][250:500])  # Convert to list

    # Repeat until all symbols succeed
    while len(symbols_to_retry) > 0:
        error_symbols = []  # Reset for each attempt

        with concurrent.futures.ThreadPoolExecutor(max_workers=4) as executor:
            futures = [executor.submit(fetch_data, symbol, error_symbols)
                       for symbol in symbols_to_retry]

            # Optionally, wait for all threads to complete
            for future in concurrent.futures.as_completed(futures):
                try:
                    future.result()  # This will raise any exceptions caught in the thread
                except Exception as exc:
                    print(f"Thread raised an exception: {exc}")

        # Update the list with symbols that encountered errors in this run
        symbols_to_retry = error_symbols

        if len(symbols_to_retry) > 0:
            print(f"\nRetrying {len(symbols_to_retry)} symbols with errors...\n")
        else:
            print("\nAll symbols processed successfully without errors.\n")

# Timing the execution
start_time = time.time()  # Start time
run_in_threads()           # Run the function
end_time = time.time()   # End time
decide_file=temp_file
tim1_decide_file=tim1_data
tim1_data=pd.DataFrame()
temp_file=pd.DataFrame()

# Calculate and print the total time taken
total_time = end_time - start_time
print(f"\nTotal time taken: {total_time:.2f} seconds")
```
