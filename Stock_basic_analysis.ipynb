```python
def men(data):
    data['diff']=data['High']-data['Low']
    data['per']=(data['diff']/data['Low'])*100
    data['type']=np.where(data['Open']<data['Close'],"up", "down") # to know wheather the stock rised (up) or fall(down)
    data['rsi14']=ta.rsi(data['Close'],lenght=14) #RSI with length of 14 
    data['obv']=ta.obv(close=data['Close'],volume=data['Volume'])
    data['ema20']=ta.ema(close=data['Close'],length=20)
    data[["bbl","bbm","bbu","bbb","bbp"]]=(ta.bbands(close=data['Low'],mamode='ema'))
    data['adx']=ta.adx(close=data['Close'],high=data['Open'],low=data['Low'],length=14)['ADX_14']
    a=data['per'][-200:-1].mean()
    return data
```
